```python
pizza_sales=df.groupby('pizza_name')['quantity'].sum().sort_values(ascending=False)
pizza_sales.head(10)
```

```python
size_sales=df.groupby('pizza_size')['quantity'].sum()
size_sales
```

# Data Visualization

```python
top_10=pizza_sales.head(10)
plt.figure(figsize=(12,6))
plt.bar(top_10.index,top_10.values)
plt.xlabel("pizza name")
plt.ylabel("quantity sold")
plt.title('top 10 most sold pizzas')
plt.xticks(rotation=45,ha="right")
plt.tight_layout()
plt.show
```

```python
df['month']=df['order_date'].dt.month
```

```python
df['total_price'] = pd.to_numeric(df['total_price'], errors='coerce')
pizza_revenue = df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=False)
top_5_revenue = pizza_revenue.head(5)
print(top_5_revenue)
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_5_revenue.index, top_5_revenue.values)
plt.gca().set_xlabel("Pizza Name")
plt.gca().set_ylabel("Revenue")
plt.gca().set_title("Top 5 Pizzas by Revenue")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
```
